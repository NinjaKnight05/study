```python
import spacy
!python -m spacy download en_core_web_sm
nlp = spacy.load("en_core_web_sm")
```

```python
text = "Apple is looking at buying a startup in India for $1 billion"
```

```python
doc = nlp(text) # It runs a pipeline including tokenization, POS tagging, lemmatization, and NER to convert raw text into structured linguistic features
```

```python
doc
```

```python
#tokenization

for token in doc:
    print(token.text)
```

```python
#Lemmatization and POS
for token in doc:
    print(f"{token.text}-->{token.lemma_}-->{token.pos_}")
```

```python
#named entity recoganization (NER)
for ent in doc.ents:
    print(f"{ent.text}-->{ent.label_}")
```

```python
#stopwords
filtered_tokens = [token.text for token in doc if not token.is_stop]
```

```python
filtered_tokens
```

```python
clean_text = " ".join([token.lemma_ for token in doc if not token.is_stop and token.is_alpha])
print("\nCLEAN TEXT:")
print(clean_text)
```

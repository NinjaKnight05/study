```python
import nltk
```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
from nltk.corpus import stopwords
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
# nltk.download('punkt_tab')
```

```python
text = "I am learning Natural Language Processing and it is very interesting"
```

```python
# tokenization part
words = word_tokenize(text)
```

```python
words
```

```python
# stop words like is am are 
new_list = [w for w in words if w.lower() not in stopwords.words('english')]
```

```python
new_list
```

```python
#stemming 
ps = PorterStemmer()
```

```python
stemmed = [ps.stem(w) for w in new_list]
```

```python
stemmed
```

```python
print("Tokens:", words)
print("After Stopwords:", new_list)
print("After Stemming:", stemmed)
```

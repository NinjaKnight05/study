```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag
```

```python
text1 = "The children are running fast"

words1 = word_tokenize(text1)
```

```python
words1
```

```python
from nltk.corpus import stopwords
filtered = [w for w in words1 if w.lower() not in stopwords.words('english')]
filtered
```

```python
pos_tags = pos_tag(filtered)
print(pos_tags)
```

```python
from nltk.corpus import wordnet
```

```python
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lem = WordNetLemmatizer()
lematized = [lem.lemmatize(word,get_wordnet_pos(tag))
             for word ,tag in pos_tags
             ]
```

```python

print("Filtered:", filtered)
print("POS Tags:", pos_tags)
print("Lemmatized:", lematized)
```

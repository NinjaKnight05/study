```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import joblib
```

```python

# =========================
# Load data
# =========================
df = pd.read_csv('car_model_rows.csv')
df
```

```python

df.rename(columns={'Selling_Price': 'target'}, inplace=True)
df.rename(columns={'Year': 'Car_age'}, inplace=True)

# Drop unused column
df.drop('Owner', axis=1, inplace=True)
```

```python
df.head()
```

```python

# Encoding (FIXED)
# =========================
# 🔧 CHANGED: separate encoders (no overwriting)
fuel_enc = LabelEncoder()
seller_enc = LabelEncoder()
trans_enc = LabelEncoder()

df['Fuel_Type'] = fuel_enc.fit_transform(df['Fuel_Type'])
df['Seller_Type'] = seller_enc.fit_transform(df['Seller_Type'])
df['Transmission'] = trans_enc.fit_transform(df['Transmission'])

```

```python
CURRENT_YEAR = 2025
df['car_age'] = CURRENT_YEAR - df['Car_age']

# 🔧 CHANGED: clean drops
df.drop(['Car_age', 'Car_Name'], axis=1, inplace=True)

# Optional column ordering (kept)
df = df[['car_age'] + [c for c in df.columns if c != 'car_age']]

# Save processed data
df.to_csv("car_price_updated.csv", index=False)
```

```python


# =========================
# Feature / Target split
# =========================
# 🔧 CHANGED: remove Present_Price (DATA LEAKAGE FIX)
X = df.drop(['target', 'Present_Price'], axis=1)
y = df['target']

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=32
)

# =========================
# Model
# =========================
model = RandomForestRegressor(
    n_estimators=110,
    random_state=42,
    max_depth=4  # you can test 7 later
)

model.fit(x_train, y_train)

# =========================
# Evaluation
# =========================
pred = model.predict(x_test)

print("Train R2:", model.score(x_train, y_train))   # 🔧 CHANGED
print("Test R2 :", model.score(x_test, y_test))     # 🔧 CHANGED
print("MAE     :", mean_absolute_error(y_test, pred))
print("MSE     :", mean_squared_error(y_test, pred))
print("R2      :", r2_score(y_test, pred))


# =========================
# Save model
# =========================
# joblib.dump(model, 'car_model.joblib')

# # =========================
# # Test prediction (FIXED)
# # =========================
# dt = joblib.load('car_model.joblib')

# # 🔧 CHANGED: year → car_age, feature order fixed
# data = [[
#     2025 - 2004,   # car_age
#     71097,         # Kms_Driven
#     0,             # Fuel_Type
#     1,             # Seller_Type
#     1              # Transmission
# ]]

# result = dt.predict(data)
# result
```

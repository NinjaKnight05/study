```python
import pandas as pd
```

```python
df = pd.read_csv('income.csv')
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
scaler = MinMaxScaler()
```

```python
df.loc[:,['Age,','Income($)']]= scaler.fit_transform(df.loc[:,['Age','Income($)']])
```

```python
import matplotlib.pyplot as plt
plt.scatter(x= df.loc[:,'Age'],y=df.loc[:,'Income($)'])
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3)
```

```python
target = model.fit_predict(df.loc[:,['Age','Income($)']])
```

```python
target
```

```python
df.loc[:,'target'] = target
```

```python
df
```

```python
model.cluster_centers_
```

```python
center = model.cluster_centers_
```

```python
center[:,0]
```

```python
df[df.loc[:,'target']==0]['Age']
```

```python
plt.scatter(df[df.loc[:,'target']==0]['Age'], df[df.loc[:,'target']==0]['Income($)'])
plt.scatter(df[df.loc[:,'target']==1]['Age'], df[df.loc[:,'target']==1]['Income($)'])
plt.scatter(df[df.loc[:,'target']==2]['Age'], df[df.loc[:,'target']==2]['Income($)'])
plt.scatter(center[:,0],center[:,1], marker='*')
```

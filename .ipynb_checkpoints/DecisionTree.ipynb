```python
import pandas as pd
```

```python
df=pd.read_csv('diabetes.csv')
```

```python
df.head(10)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
```

```python
enc = LabelEncoder()
```

```python
gender_enc=enc.fit_transform(df.gender)
smoking_enc=enc.fit_transform(df.smoking_history)
```

```python
df.insert(loc=0, column="Gender_enc", value = gender_enc)
df.insert(loc=4, column="Smoking_hist_enc", value = smoking_enc)
```

```python
df.head()
```

```python
df.smoking_history.value_counts()
```

```python
df.Smoking_hist_enc.value_counts()
```

```python
df.drop(columns = ['gender', 'smoking_history'], inplace = True)
```

```python
df.diabetes.value_counts()
```

```python
df.rename(columns={'diabetes' : 'target'}, inplace = True)
```

```python
df.head()
```

```python
x = df.iloc[:, :-1]
```

```python
y = df.iloc[:,-1]
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2,random_state = 35)
```

```python
x_train
```

```python
y_train
```

```python
dt = DecisionTreeClassifier()
```

```python
dt.fit(x_train,y_train)
```

```python
y_test
```

```python
pred = dt.predict(x_test)
```

```python

accuracy_score(y_test,pred)
```

```python
import joblib
joblib.dump(dt, 'diabetes_model.pkl')
```

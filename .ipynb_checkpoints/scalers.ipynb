```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('income.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python

df['Income($)'].min()
```

```python
df['Income($)'].max()
```

```python
st_scaler=StandardScaler()
```

```python
df.loc[:,'Standard_new']=st_scaler.fit_transform(df.loc[:,['Income($)']])
```

```python
df.head()
```

```python
mic_Scaler=MinMaxScaler()
```

```python
df.loc[:,'New_minmax']=mic_Scaler.fit_transform(df.loc[:,['Income($)']])
```

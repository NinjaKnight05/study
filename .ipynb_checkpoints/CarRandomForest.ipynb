```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import pickle
from sklearn.preprocessing import StandardScaler


```

```python
df = pd.read_csv('car_model_rows.csv')
df
```

```python
df.drop('Seller_Type',axis=1,inplace=True)
```

```python
# df[df['target']==22.31]
```

```python
# df['car_age'].min()
```

```python
df['Kms_Driven'].max()
```

```python
df.rename(columns={'Selling_Price':'target'},inplace=True)
df.rename(columns={'Year':'Car_age'},inplace=True)
```

```python
enc = LabelEncoder()
```

```python
fuel=enc.fit(df['Fuel_Type'])
with open('fuel.pkl',"wb") as f:
    pickle.dump(fuel,f)
owner=enc.fit(df['Owner'])
with open('own.pkl',"wb") as f:
    pickle.dump(owner,f)
trans=enc.fit(df['Transmission'])
with open('trans.pkl',"wb") as f:
    pickle.dump(trans,f)
```

```python
df['Fuel_Type']=enc.fit_transform(df['Fuel_Type'])
# df['Seller_Type']=enc.fit_transform(df['Seller_Type'])
df['Transmission']=enc.fit_transform(df['Transmission'])
```

```python
df.head(20)
```

```python
CURRENT_YEAR = 2025
df['car_age'] = CURRENT_YEAR - df['Car_age']
```

```python
# df.drop('ar_age',axis=1,inplace=True)
df.drop('Car_age',axis=1,inplace=True)
df.drop('Car_Name',axis=1,inplace=True)
df.head()
```

```python
# # df = df[['car_age'] + [c for c in df.columns if c != 'car_age']]
# df.to_csv("car_price_updated.csv", index=False)
# df.head()

```

```python
s_scaler = StandardScaler()
```

```python
df.loc[:,'New_kms'] = s_scaler.fit_transform(df.loc[:,['Kms_Driven']])
```

```python
df.loc[:,'Car_age'] = s_scaler.fit_transform(df.loc[:,['car_age']])
```

```python
df
```

```python
df.drop('Kms_Driven',axis=1,inplace = True)
```

```python
df.drop('car_age',axis=1,inplace = True)
```

```python
df = df[['Car_age','Present_Price', 'New_kms','Fuel_Type', 'Owner','Transmission','target']]
```

```python
x = df.iloc[:,:-1]
```

```python
x
```

```python
y=df['target']
```

```python
y
```

```python
x
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=32)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model = RandomForestRegressor(n_estimators=110,random_state=42,max_depth=4)
```

```python
model.fit(x_train,y_train)
```

```python
pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
r2_score(y_test,pred)
```

```python
mean_squared_error(y_test,pred)
```

```python
import joblib
```

```python
joblib.dump(model,'car_model.joblib')
```

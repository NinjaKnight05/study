```python
income=pd.read_csv('income.csv')
```

```python
income.head()
```

```python
income['Income($)'].min()
```

```python
income['Income($)'].max()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
st_scaler=StandardScaler()
```

```python
income.loc[:,'Income($)']
```

```python
income.loc[:,'Income($)'].std()
```

```python
# standard deviation formula
s=income.loc[:,'Income($)'].std()
u=income.loc[:,'Income($)'].mean()

z=(9000-u)/s
z
```

```python
income.loc[:,'new_Standard']=st_scaler.fit_transform(income.loc[:,['Income($)']])
```

```python
income.head()
```

Minmax Scaler


```python
minmax_scaler=MinMaxScaler(feature_range=(0,4))
```

```python
income.loc[:,'Minmax_values']=minmax_scaler.fit_transform(income.loc[:,['Income($)']])
```

```python
income.head()
```

Linear Regression


```python
sf=pd.read_csv('bmi.csv')
```

```python
sf.head()
```

```python
sf = sf.drop('BmiClass',axis=1)
```

```python
sf.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
x=sf[['Age','Height','Weight']]
x
```

```python
y=sf['Bmi']
y
```

```python
xtrain,xtest,ytrain,ytest =train_test_split(x,y,test_size=0.2,random_state=30)
```

```python
model =LinearRegression()
```

```python
model.fit(xtrain,ytrain)
```

```python
pred =model.predict(xtest)
pred
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
r2_score(ytest,pred)
```

```python
mean_squared_error(ytest,pred)
```

```python
import joblib
```

```python
joblib.dump(model,'bmi_model.joblib')   
# same-------------bmi_model.pkl)
```

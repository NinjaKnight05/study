```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

```python
# features >> independent variable
# target >> dependent variable
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
enc=LabelEncoder()
```

```python
df.Sex.value_counts()
```

```python
enc.fit(df.loc[:,'Sex'])
```

```python
ans=enc.transform(df.loc[:,'Sex'])
ans
```

```python
df["Sex"]= ans
```

```python
cf=pd.read_csv('Flights.csv')
```

```python
cf.head()
```

```python
cf.Destination.value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder()
```

```python
encoder.fit(cf.loc[:,['Destination']])
```

```python
destinations=encoder.transform(cf.loc[:,['Destination']]).toarray()
```

```python
destinations
```

```python
encoder.get_feature_names_out()
```

```python
cf.loc[:,encoder.get_feature_names_out()]= destinations
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
# from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('housing.csv')
```

```python
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df['total_bedrooms'] = df['total_bedrooms'].fillna(df['total_bedrooms'].mean())
```

```python
df.isna().sum()
```

```python
enc = LabelEncoder()
```

```python
df['Ocean_proximity'] = enc.fit_transform(df['ocean_proximity'])
```

```python
df.head()
```

```python
df.drop('ocean_proximity',axis =1 ,inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
x = df.drop('median_house_value',axis=1)
```

```python
x
```

```python
scale= sc.fit_transform(x)
```

```python
scale
```

```python
y = df['median_house_value']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(scale,y,random_state=43,test_size=0.2)
```

```python
model = RandomForestRegressor(max_depth=15,random_state=43,n_estimators=20)
```

```python
model.fit(x_train,y_train)
```

```python
pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score,confusion_matrix
```

```python
r2_score(y_test,pred)*100
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model2 = DecisionTreeRegressor(max_depth=15,random_state=42)
```

```python
model2.fit(x_train,y_train)
```

```python
pred1 = model2.predict(x_test)
```

```python
r2_score(y_test,pred1)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model3 = LinearRegression()
```

```python
model3.fit(x_train,y_train)
```

```python
pred2 = model3.predict(x_test)
```

```python
r2_score(y_test,pred2)
```

```python
import seaborn as sns
df3 = sns.load_dataset('titanic')
df3.head()
```

```python
df3.head()
```

```python
df3.drop('class',axis=1,inplace=True)
```

```python
df3
```

```python
# dummies = pd.get_dummies(x.sex)
# dummies.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
enc = OneHotEncoder()
```

```python
encoder = enc.fit_transform(df3[['sex']]).toarray()
```

```python
encoder
```

```python
df3.loc[:,enc.get_feature_names_out()] = encoder
```

```python
df3.head()
```

```python

df3.drop('sex',axis=1,inplace=True)
```

```python
x = df3.drop('survived',axis=1)
```

```python
x.columns[x.isna().any()]
```

```python
x.age[:10]
```

```python
x
```

```python
y = df3['survived']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model = GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
# model.score(x_test,y_test)
```

```python
pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
# df.loc[5989,'rating']='G'
# df.loc[6827,'rating']='PG'
# df.loc[7537,'rating']='TY-Y'
# df.loc[7312,'rating']='NR'
df['rating']=df['rating'].fillna('Not Rated')
```

```python
df['cast']=df['cast'].fillna('unknown')
```

```python
df.isna().sum()
```

```python
df['country']=df['country'].fillna('unknown')
df['duration']=df['duration'].fillna('unknown')
```

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce')
```

```python
# df.isna().sum
df.isna().sum()
```

```python
df.loc[df['date_added'].isna(), 'date_added'].head(10)
```

```python
df.info()
```

```python
df = df.dropna(subset=['show_id'])

```

```python

import plotly.express as px
```

```python
null_counts = df.isna().sum()

px.bar(
    x=null_counts.index,
    y=null_counts.values,
    title='Missing Values per Column'
).show()

```

```python

px.pie(
    names=df['type'].unique(), 
    values=df['type'].value_counts()[df['type'].unique()],
    title='Movies vs TV Shows Count'
)

```

```python
top10 = df['country'].str.split(',').str[0].value_counts().head(10)

px.bar(
    x=top10.index,       
    y=top10.values,     
    color=top10.index,  
    title='Top 10 Countries by Content'
).show()
```

```python
rating_counts = df['rating'].value_counts()

px.scatter(
    x=rating_counts.index,
    y=rating_counts.values,
    title='Content Rating Distribution'
).show()
```

```python

genre_series = df['listed_in'].str.split(', ', expand=True).stack()
genre_counts = genre_series.value_counts()

px.line(
    x=genre_counts.index,
    y=genre_counts.values,
    title='Genre Popularity'
).show()
```

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

```python
df.head(20)
```

```python
df.drop(['Name','Ticket','PassengerId','Cabin'],axis=1,inplace=True)
```

```python
df['Embarked'].unique()
```

```python
# WRONG for Bernoulli
# from sklearn.preprocessing import LabelEncoder
# df["Embarked_le"] = LabelEncoder().fit_transform(df["Embarked"])
```

```python
nulls = df.columns[df.isna().any()]
nulls
```

```python
df.isna().sum()
```

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
```

```python
df.head()
```

```python
df['Embarked'].fillna(df['Embarked'].mode()[0],inplace=True)
```

```python
df.isna().sum()
```

```python
# from sklearn.preprocessing import Binarizer

# bin = Binarizer(threshold=30)
# df["Age_bin"] = bin.fit_transform(df[["Age"]])
```

```python
df["IsChild"] = (df["Age"] < 18).astype(int)
df["IsAdult"] = ((df["Age"] >= 18) & (df["Age"] < 60)).astype(int)
df["IsSenior"] = (df["Age"] >= 60).astype(int)
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=["Embarked"], drop_first=True)
```

```python
df["Sex"] = df["Sex"].map({"male":0, "female":1})
```

```python
df.head()
```

```python
df["Fare_bin"] = (df["Fare"] > df["Fare"].median()).astype(int)
```

```python
df.head()
```

```python
df.drop(['Age','Fare'],axis=1,inplace=True)
```

```python
df.head()
```

```python
x = df.drop('Survived',axis=1)
```

```python
x
```

```python
y = df['Survived']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.naive_bayes import BernoulliNB
```

```python
model = BernoulliNB()
```

```python
model.fit(xtrain,ytrain)
```

```python
pred = model.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
```

```python
accuracy_score(pred,ytest)*100
```

```python
cm=confusion_matrix(pred,ytest)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(3,2))
sns.heatmap(cm,annot=True,cmap='summer')
plt.xlabel("Prediction")
plt.ylabel("Truth Values")
plt.title("Confusion Matrix")
plt.show()
```

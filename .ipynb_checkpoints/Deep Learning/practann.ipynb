```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("diabetes.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.diabetes.value_counts()
```

```python
df = pd.get_dummies(df,columns=['smoking_history','gender'],drop_first=True)
# df = pd.get_dummies(df,columns=['smoking_history','gender'])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
x = df.drop(columns='diabetes')
y = df.diabetes
```

```python
x
```

```python
y
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
sc = StandardScaler()
```

```python
xtrain_Scaled =sc.fit_transform(xtrain)
xtest_scaled = sc.transform(xtest)
```

```python
xtrain_Scaled.shape
```

```python
import tensorflow
```

```python
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(15,activation='relu',input_dim = 13))
model.add(Dense(15,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
model.summary()
```

```python
epo = model.fit(xtrain_Scaled,ytrain,epochs=20,validation_split=0.2)
```

```python
model.layers[0].get_weights()
```

```python
pred = model.predict(xtest_scaled)
```

```python
ypred = np.where(pred > 0.5 ,1 ,0 )
```

```python
ypred
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
```

```python
accuracy_score(ytest,ypred)*100
```

```python
epo.history
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(epo.history['loss'],label='Traing Loss')
plt.plot(epo.history['val_loss'],label = 'Validation Loss')
plt.legend()
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
plt.plot(epo.history['accuracy'])
plt.plot(epo.history['val_accuracy'])
```

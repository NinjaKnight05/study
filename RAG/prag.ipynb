```python
from langchain_community.document_loaders import PyPDFLoader
loding = PyPDFLoader('25 feb research paper.pdf')
doc1 = loding.load()
```

```python
doc1
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
split = RecursiveCharacterTextSplitter(chunk_size=800,chunk_overlap=200)
```

```python
document = split.split_documents(doc1)
```

```python
document
```

```python
def cleantext(text):
    return text.encode('utf-8','ignore')
cleaned = []
for i in document:
    i.page_content=cleantext(i.page_content)
    cleaned.append(i)
```
